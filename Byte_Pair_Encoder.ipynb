```python
import tiktoken
tokenizer = tiktoken.get_encoding('gpt2')
text = "Transformers are deep neural networks that power llm"
print(text)
token_ids = tokenizer.encode(text)
print(token_ids)
for i in token_ids:
  print("Token for " + str(i) + " is :")
  print(tokenizer.decode([i]))
```
