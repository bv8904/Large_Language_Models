```python
#Opening the text file and copying its contents in text variable
with open('/content/SherlockHolmes.txt','r', encoding='utf-8')as f:
  text = f.read()
print(f"Length of text : {len(text)}")
print()
print(text[:50])
print()


import re
#Splitting the text into tokens
result = re.split(r'([?.,;"@#]|--|\s)',text )
print(f"The tokens are: {result[:50]}")
print()

#Removing whitespace as a token
result = [item.strip() for item in result if item.strip()]
print(f"The tokens without spaces are: {result[:50]}")
print()

#Sorting the unique tokens alphabetically
unique_tokens = sorted(set(result))
print(f"The length of vocabulary is {len(unique_tokens)}")
print()

#Adding special characters to our unique tokens list
unique_tokens.extend(['<unk>', '<pad>'])
print(f"The new length of vocabulary {len(unique_tokens)}")
print()

#Creating the vocab dictionary mapping token to ID
vocab = {token: id for id, token in enumerate(unique_tokens)}

#Printing dictionary
print(list(vocab.items())[:20])
```
