```python
#Importing tiktoken
import tiktoken
#Instantiating
tokenizer = tiktoken.get_encoding('gpt2')

text = "Transformers are deep neural networks that power llm"
print(text)
print()

#Tokens to token ID's
token_ids = tokenizer.encode(text)
print(f"The Token ID's are {token_ids}")
print()

for i in token_ids:
  print(f"Token for ID {i} is : ")
  #Token ID's to tokens
  print(tokenizer.decode([i]))
```
