```python
import torch
import matplotlib.pyplot as plt
```

```python
inputs = torch.tensor([
   [0.43, 0.15, 0.89], # A 3D Vector Embedding for token - Your
   [0.55, 0.87, 0.66], # A 3D Vector Embedding for token - journey
   [0.57, 0.85, 0.64], # A 3D Vector Embedding for token - starts
   [0.22, 0.58, 0.33], # A 3D Vector Embedding for token - with
   [0.77, 0.25, 0.10], # A 3D Vector Embedding for token - one
   [0.05, 0.80, 0.55]] # A 3D Vector Embedding for token - step
                       )
```

```python
d_in = inputs.shape[1]
d_out = 2
torch.manual_seed(123)
#Trainable query weight matrix
W_query = torch.nn.Parameter(torch.rand(d_in, d_out), requires_grad=False)
#Trainable key weight matrix
W_key = torch.nn.Parameter(torch.rand(d_in, d_out), requires_grad=False)
#Trainable value weight matrix
W_value = torch.nn.Parameter(torch.rand(d_in, d_out), requires_grad=False)
print(W_query)
print(W_key)
print(W_value)
```

```python
#Key matrix
key = inputs @ W_key
#Query matrix
query = inputs @ W_query
#Value matrix
value = inputs @ W_value
print(key)
print(f"Dimensions of key matrix is {key.shape}")
print(query)
print(f"Dimensions of query matrix is {query.shape}")
print(value)
print(f"Dimensions of value matrix is {value.shape}")
```

```python
#Computing attention score
attention_scores = query @ key.T
print(attention_scores)
```

```python
d_k=key.shape[-1]
#Computing attention weights
attention_weights = torch.softmax(attention_scores/d_k**0.5,dim=-1)
print(attention_weights)
```

```python
#Computing context vector
context_vector = attention_weights @ value
print(context_vector)
```
